```python

print("===== STUDENT MARKS ANALYSER =====")

name = input("Enter student name: ")

subjects = ["English", "Maths", "Python", "DBMS", "Java"]
marks = []

for subject in subjects:
    score = float(input("Enter marks in " + subject + ": "))
    marks.append(score)

total = sum(marks)
percentage = total / len(subjects)

highest = max(marks)
lowest = min(marks)

if lowest >= 40:
    result = "PASS"
else:
    result = "FAIL"

if percentage >= 90:
    grade = "A+"
elif percentage >= 80:
    grade = "A"
elif percentage >= 70:
    grade = "B"
elif percentage >= 60:
    grade = "C"
elif percentage >= 40:
    grade = "D"
else:
    grade = "F"

print("\n===== STUDENT REPORT =====")
print("Student Name:", name)
print("Total Marks:", total, "/ 500")
print("Percentage:", round(percentage, 2), "%")
print("Highest Marks:", highest)
print("Lowest Marks:", lowest)
print("Grade:", grade)
print("Final Result:", result)

print("==========================")
```
